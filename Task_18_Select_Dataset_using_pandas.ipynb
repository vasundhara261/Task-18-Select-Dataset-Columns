```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
# Load the Iris dataset
iris = load_iris()

# Create a Pandas DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add the target column
df["target"] = iris.target

# Display the first 5 rows
print("First 5 rows of the dataset:")
print(df.head())
```

```python
# Select all rows and specific columns
selected_loc = df.loc[:, ["sepal length (cm)", "petal length (cm)"]]

print("Selected columns using loc:")
print(selected_loc.head())
```

```python
# Select rows 0 to 4 and specific columns
result_loc = df.loc[0:4, ["sepal length (cm)", "sepal width (cm)"]]

print("Rows and columns selected using loc:")
print(result_loc)
```

```python
# Select the first 5 rows and first 2 columns
selected_iloc = df.iloc[0:5, 0:2]

print("Selected data using iloc:")
print(selected_iloc)
```

```python
# Select a single column
single_column = df["sepal length (cm)"]

print("Single column:")
print(single_column.head())
```

```python
# Select multiple columns
multiple_columns = df[["sepal width (cm)", "petal width (cm)"]]

print("Multiple columns:")
print(multiple_columns.head())
```
